```python
from tensorflow import keras
(train_input, train_target), (test_input, test_target) = keras.datasets.fashion_mnist.load_data()
```

```python
from sklearn.model_selection import train_test_split
train_scaled = train_input / 255.0
train_scaled = train_scaled.reshape(-1, 28*28)
train_scaled, val_scaled, train_target, val_target = train_test_split(train_scaled, train_target, test_size = 0.2, random_state = 42)
```

```python
# hidden layer: all the dense layer btw input layer and output layer
# hidden layer has their own activation function
# unlike output layer, hidden layer is more free to choose activation function
# most popular activation functions are sigmoid and ReLU
```

```python
# why activation function on hidden layer?
# if two equation is linearly connected like (a * 4 + 2 = b) and (b * 3 - 5 = c), actually b is useless since we can change it to [(a * 4 + 2) * 3 - 5 = c]
# it just like to hidden layer
# so we need to twist linear calculation to kinda unlinear calculation such as (a * 4 + 2 = b), (log(b) = k), and (k * 3 - 5 = c)
```

```python
dense1 = keras.layers.Dense(100, activation='sigmoid', input_shape=(784,))          # first layer need an input_shape
dense2 = keras.layers.Dense(10, activation='softmax')
# # of neurons of hidden layer is free to choose but it has to bigger than neuron of output layer (classes)
```

```python
model = keras.Sequential([dense1, dense2])
# output layer is on the last of list
```

```python
model.summary()
```

```python
# name(): change the name of dense layer
# Output Shape (None, 100): None = sample size which is undefined since fit() method of keras use minibatch gradient descent and do not fix the samplsize and let by None to respond fkexibility to minibatch size
# Param # 78500 : 784 (pixels) * 100 (neurons weight) + 100 (neurons intercept) = 78500
# Param # 1010 : 100 (prev neuron) * 10 (subsequence neuron weight) + 10 (sub neuron weight) = 1010
```

```python
# different way to add layer
model = keras.Sequential([
    keras.layers.Dense(100, activation='sigmoid', input_shape=(784,), name = 'hidden'),
    keras.layers.Dense(10, activation='softmax', name='output')
], name = 'Fashion MNIST Model')
```

```python
model.summary()
```

```python
# most use way to add layer
model = keras.Sequential()
model.add(keras.layers.Dense(100, activation = "sigmoid", input_shape=(784,)))
model.add(keras.layers.Dense(10, activation='softmax'))
```

```python
model.summary()
```

```python
model.compile(loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model.fit(train_scaled, train_target, epochs=5)
```

```python
# .reshape(-1, 28*28) ==> Flatten Class from keras
# Flatten layer is just flat the input to 1 dimension just like reshape()
# Flatten layer is btw input layer and hidden layer
```

```python
model = keras.Sequential()
model.add(keras.layers.Flatten(input_shape=(28, 28)))
model.add(keras.layers.Dense(100, activation='relu'))
model.add(keras.layers.Dense(10, activation='softmax'))
```
